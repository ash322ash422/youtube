# 1) Hugging Face - Embedding Models
AKA Sentence transformers

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

# 1. Load a lightweight, high-quality embedding model: 120 MB
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
# 2. Define our documents (text to convert into vectors)
documents = [
    "How to reset your account password.",
    "The weather is very sunny and clear today.",
    "Updating user login credentials and security phrases."
]

# 3. Convert the documents into vector embeddings:
# An array of 384 floating-point numbers
doc_vectors = model.encode(documents)

print("--- Vector Preview ---")
print(f"Document 1 vector shape: {doc_vectors[0].shape}")  # Output: (384,)
print(f"First 20 numbers of Document 1 vector:\n{doc_vectors[0][:20]}\n")
```

```python
# 4. Process a user search query
query = "I forgot how to log into my profile"
query_vector = model.encode([query])  # Convert query to a vector

# 5. Calculate similarity (simulate what a Vector DB does)
# This measures the mathematical angle between the vectors
similarities = cosine_similarity(query_vector, doc_vectors)[0]

print("--- Search Results (Semantic Match) ---")
for i, score in enumerate(similarities):
    print(f"Score: {score:.4f} | Document: '{documents[i]}'")
```

Document 1 and 3 are similar to the query

# 2) OpenAI - Embedding Model

```python
from openai import OpenAI
import textwrap

# 1. Initialize the OpenAI client
OPENAI_API_KEY = "key_here"
client = OpenAI(api_key=OPENAI_API_KEY)

# 2. Define the text we want to convert into a vector
text_to_embed = "How to reset your account password."

# 3. Call the OpenAI Embedding API
# We use 'text-embedding-3-small', which is fast, accurate, and cheap
response = client.embeddings.create(
    model="text-embedding-3-small", 
    input=text_to_embed)

# 4. Extract the mathematical vector array from the response
vector = response.data[0].embedding
vector_string = str(vector[:20])
wrapped_text = textwrap.fill(vector_string, width=80)

print("--- OpenAI Vector Details ---")
print(f"Vector Dimensions: {len(vector)}")  # Output: 1536 dimensions
print(f"First 20 numbers of the OpenAI vector:\n{wrapped_text}")
```
